COUNT PLOT (Compare the number of movies by genre.)


```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

dmd = pd.read_csv('/content/netflix_titles.csv', encoding='latin1')
genre = dmd['listed_in'].str.split(', ').explode()

plt.figure(figsize=(12,6))
sns.countplot(y=genre.reset_index(drop=True), order=genre.value_counts().index[:10])

plt.title("Top 10 Genres")
plt.xlabel("Number of Movies")
plt.ylabel("Genre")
plt.show()
```

HISTOGRAM(Release Year Distribution.)

```python
plt.figure(figsize=(10,5))

sns.histplot(dmd['release_year'], bins=20, kde=True)

plt.title("Distribution of Release Years")
plt.xlabel("Release Year")
plt.ylabel("Count")

plt.show()
```

SCATTER PLOT(Alternative: Release Year vs Duration)

```python
movies = dmd[dmd['type']=="Movie"].copy()

movies['duration'] = movies['duration'].str.replace(' min','')
movies['duration'] = pd.to_numeric(movies['duration'], errors='coerce')

plt.figure(figsize=(10,6))

sns.scatterplot(data=movies,
                x='release_year',
                y='duration')

plt.title("Release Year vs Movie Duration")
plt.xlabel("Release Year")
plt.ylabel("Duration (minutes)")

plt.show()
```

BOX PLOT(Movie Duration by Rating)

```python
movies = dmd[dmd['type']=="Movie"].copy()
movies['duration'] = movies['duration'].str.replace(' min','')
movies['duration'] = pd.to_numeric(movies['duration'], errors='coerce')

plt.figure(figsize=(12,6))

sns.boxplot(data=movies,
            x='rating',
            y='duration')

plt.xticks(rotation=45)

plt.title("Movie Duration by Rating")
plt.xlabel("Rating")
plt.ylabel("Duration (minutes)")

plt.show()
```

HEATMAP (Correlation)

```python
movies = dmd[dmd['type']=="Movie"].copy()
movies['duration'] = movies['duration'].str.replace(' min','')
movies['duration'] = pd.to_numeric(movies['duration'], errors='coerce')

corr = movies[['release_year','duration']].corr()

plt.figure(figsize=(6,4))

sns.heatmap(corr,
            annot=True,
            cmap='coolwarm',
            linewidths=1)

plt.title("Correlation Heatmap")

plt.show()
```
